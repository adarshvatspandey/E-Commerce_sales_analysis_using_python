# Advanced Questions

# 21. RFM Customer Segmentation

```python
snapshot_date = data['Order Date'].max()

rfm = data.groupby('Customer ID').agg({
    'Order Date': lambda x: (snapshot_date - x.max()).days,
    'Order ID': 'nunique',
    'Sales': 'sum'
})

rfm.columns = ['Recency','Frequency','Monetary']

rfm.head()
```

```python
Insight:Identifies loyal and at-risk customers.
```

# 22. Pareto Analysis (80-20 Rule)

```python
customer_sales = (
    data.groupby('Customer Name')['Sales']
        .sum()
        .sort_values(ascending=False)
)

(customer_sales.cumsum()/customer_sales.sum())*100
```

```python
Insight:Checks if 20% customers generate 80% revenue.
```

# 23. Customer Retention Analysis

```python
repeat_customers = (
    data.groupby('Customer ID')['Order ID']
        .nunique()
)

repeat_customers[repeat_customers > 1].count()
```

# 25. Profit Margin Analysis

```python
profit_margin = (
    data['Profit'].sum() /
    data['Sales'].sum()
) * 100

print(profit_margin)
```
